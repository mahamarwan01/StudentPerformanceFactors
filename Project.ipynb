```python

def Desc_stat(ds, var):
    mean = ds[var].mean()
    median = ds[var].median()
    mode = ds[var].mode()
    minimum = ds[var].min()
    maximum = ds[var].max()
    range = maximum - minimum
    std_dev = ds[var].std()
    variance = ds[var].var()
    skew = ds[var].skew()
    kurtosis = ds[var].kurt()
    count = ds[var].count()

    summary = {
        "Average": mean,
        "Median": median,
        "Mode": mode.tolist(),
        "Minimum": minimum,
        "Maximum": maximum,
        "Range": range,
        "Std Dev": std_dev,
        "Variance": variance,
        "Skewness": skew,
        "Kurtosis": kurtosis,
        "Count": count
    }

    return summary
```

```python

def desc_stats(ds, var):
    mean = ds[var].mean()
    median = ds[var].median()
    mode = ds[var].mode()
    minimum = ds[var].min()
    maximum = ds[var].max()
    range = maximum - minimum
    std_dev = ds[var].std()
    variance = ds[var].var()
    skew = ds[var].skew()
    kurtosis = ds[var].kurt()

    Quart = [
        ds[var].quantile(0),
        ds[var].quantile(0.25),
        ds[var].quantile(0.50),
        ds[var].quantile(0.75),
        ds[var].quantile(1),
        ds[var].quantile(0.75) - ds[var].quantile(0.25)
    ]

    summary = {
        "Average": mean,
        "Median": median,
        "Mode": mode.tolist(),
        "Minimum": minimum,
        "Maximum": maximum,
        "Range": range,
        "Std Dev": std_dev,
        "Variance": variance,
        "Skewness": skew,
        "Kurtosis": kurtosis,
        "25th Percentile": Quart[1],
        "50th Percentile": Quart[2],
        "75th Percentile": Quart[3],
        "Inter Quartile Range": Quart[5],
        "Count": ds[var].count()
    }

    return summary
```

```python

# Chi-square Test
import pandas as pd
from scipy.stats import chi2_contingency
# Create categories for exam scores
data = df[["Parental_Involvement", "Exam_Score"]].dropna().copy()

data["Score_Category"] = pd.cut(
    data["Exam_Score"],
    bins=[-float("inf"), 60, 70, float("inf")],
    labels=["Low", "Medium", "High"]
)

# Create a contingency table
table = pd.crosstab(data["Parental_Involvement"], data["Score_Category"])

print("Contingency Table:")
print(table)

# Perform chi-square test
chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)

# Check the result
if p_value < 0.05:
    print("Reject the null hypothesis: There is a significant association.")
else:
    print("Do not reject the null hypothesis: There is no significant association.")
```

```python

# One-Sample T-test
from scipy.stats import ttest_1samp

# Select the dependent variable
sample = df["Exam_Score"].dropna()

# Set the hypothesised population mean
population_mean = 65

# Perform the one-sample t-test
t_stat, p_value = ttest_1samp(sample, population_mean)

print("Sample Mean:", sample.mean())
print("Hypothesised Population Mean:", population_mean)
print("T-statistic:", t_stat)
print("P-value:", p_value)

# Interpret the result
if p_value < 0.05:
    print("Reject the null hypothesis.")
    print("The sample mean is significantly different from the hypothesised mean.")
else:
    print("Do not reject the null hypothesis.")
    print("There is insufficient evidence that the sample mean differs from the hypothesised mean.")
```

```python
# Step 1 - Import libraries and load the dataset
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

dataset = pd.read_csv('StudentPerformanceFactors.csv')
print('Dataset shape:', dataset.shape)
print(dataset.head())
print(dataset.info())

features = ['Hours_Studied', 'Attendance', 'Sleep_Hours',
            'Previous_Scores', 'Tutoring_Sessions', 'Physical_Activity']
X = dataset[features]
y = dataset['Exam_Score']
print('Independent variables:', X.columns.tolist())
print('Dependent variable: Exam_Score')
```

```python
# Step 2 - Check missing values and duplicate rows
print('Missing values by column:')
print(dataset.isnull().sum())
print('Duplicate rows:', dataset.duplicated().sum())
print('Numeric descriptive statistics:')
print(dataset[features + ['Exam_Score']].describe().round(2))
```

```python
# Step 3 - Identify categorical variables for classification
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

X_class = dataset.drop(columns=['Exam_Score'])
num_cols = X_class.select_dtypes(include=np.number).columns.tolist()
cat_cols = X_class.select_dtypes(exclude=np.number).columns.tolist()
print('Numerical columns:', num_cols)
print('Categorical columns:', cat_cols)

try:
    encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
except TypeError:
    encoder = OneHotEncoder(handle_unknown='ignore', sparse=False)

preprocessor = ColumnTransformer([
    ('num', Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())]), num_cols),
    ('cat', Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', encoder)]), cat_cols)
])
```

```python
# Step 4 - Train/test split for regression
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42)
print('Training rows:', len(X_train))
print('Testing rows:', len(X_test))
```

```python
# Step 5 - Demonstrate training-only feature scaling
sc = StandardScaler()
X_train_scaled = sc.fit_transform(X_train)
X_test_scaled = sc.transform(X_test)
print('Scaled training sample:')
print(pd.DataFrame(X_train_scaled, columns=features).head().round(2))
```

```python
# Step 6 - Build all five regression models
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

reg_models = {
    'Multiple Linear Regression': LinearRegression(),
    'Polynomial Regression (degree 2)': Pipeline([
        ('poly', PolynomialFeatures(degree=2, include_bias=False)),
        ('model', Ridge(alpha=100.0))]),
    'Lasso Regression': Pipeline([
        ('scale', StandardScaler()),
        ('model', Lasso(alpha=0.05, max_iter=10000))]),
    'Ridge Regression': Pipeline([
        ('scale', StandardScaler()),
        ('model', Ridge(alpha=1.0))]),
    'Elastic Net': Pipeline([
        ('scale', StandardScaler()),
        ('model', ElasticNet(alpha=0.05, l1_ratio=0.5,
                             max_iter=10000))])
}
reg_results = []
for name, regressor in reg_models.items():
    regressor.fit(X_train, y_train)
    predicted = regressor.predict(X_test)
    reg_results.append({
        'Model': name,
        'R2': r2_score(y_test, predicted),
        'MSE': mean_squared_error(y_test, predicted),
        'MAE': mean_absolute_error(y_test, predicted)
    })

reg_table = pd.DataFrame(reg_results).sort_values('R2', ascending=False)
print(reg_table.round(4).to_string(index=False))
best_reg_name = reg_table.iloc[0]['Model']
best_reg_model = reg_models[best_reg_name]
print('Best regression model:', best_reg_name)
```

```python
# Step 7 - Compare actual and predicted results
best_predictions = best_reg_model.predict(X_test)
comparison = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': np.round(best_predictions, 2)
})
print(comparison.head(10))
plt.scatter(y_test, best_predictions, alpha=0.4)
low = min(y_test.min(), best_predictions.min())
high = max(y_test.max(), best_predictions.max())
plt.plot([low, high], [low, high], color='red', linestyle='--')
plt.xlabel('Actual Exam Score')
plt.ylabel('Predicted Exam Score')
plt.title('Actual vs Predicted: Best Regression Model')
plt.show()
```

```python
# Step 8 - Convert the Lasso equation to original predictor units
lasso_scaler = best_reg_model.named_steps['scale']
lasso_fit = best_reg_model.named_steps['model']
coefficients = lasso_fit.coef_ / lasso_scaler.scale_
intercept = lasso_fit.intercept_ - np.sum(
    lasso_fit.coef_ * lasso_scaler.mean_ / lasso_scaler.scale_)
print('Intercept:', round(intercept, 6))
for variable, coefficient in zip(features, coefficients):
    print(variable, ':', round(coefficient, 6))
print('Equation: Exam_Score =', round(intercept, 5),
      '+ SUM(coefficient * feature)')

# Select five test rows and calculate predictions by substitution
selected = X_test.sample(n=5, random_state=7)
for row_index, row in selected.iterrows():
    calculation = intercept + np.dot(coefficients, row.values)
    print('CSV row:', row_index + 2,
          'Actual:', dataset.loc[row_index, 'Exam_Score'],
          'Predicted:', round(calculation, 3),
          'Values:', row.to_dict())
```

```python
# Step 9 - Create classification target and split
from sklearn.model_selection import train_test_split

y_class = (dataset['Exam_Score'] >= 70).astype(int)
X_class = dataset.drop(columns=['Exam_Score'])
print('Full dataset class counts:')
print(y_class.value_counts().sort_index())
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_class, y_class, test_size=0.20,
    random_state=42, stratify=y_class)
```

```python
# Step 10 - Train and evaluate classification models
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.metrics import (accuracy_score, precision_score,
    recall_score, f1_score, confusion_matrix, classification_report)

classifiers = {
 'Logistic Regression': LogisticRegression(
     max_iter=3000, class_weight='balanced'),
 'K-Nearest Neighbors': KNeighborsClassifier(n_neighbors=15),
 'Support Vector Machine': SVC(kernel='rbf', class_weight='balanced'),
 'Decision Tree': DecisionTreeClassifier(
     max_depth=7, min_samples_leaf=10, random_state=42),
 'Random Forest': RandomForestClassifier(
     n_estimators=150, min_samples_leaf=3,
     random_state=42, n_jobs=-1),
 'Extra Trees (additional)': ExtraTreesClassifier(
     n_estimators=150, min_samples_leaf=3,
     random_state=42, n_jobs=-1)
}
class_results = []
fitted_models = {}
for name, classifier in classifiers.items():
    model = Pipeline([('preprocess', preprocessor),
                      ('classifier', classifier)])
    model.fit(X_train_c, y_train_c)
    y_pred = model.predict(X_test_c)
    fitted_models[name] = model
    class_results.append({
      'Model': name,
      'Accuracy': accuracy_score(y_test_c, y_pred),
      'Precision': precision_score(y_test_c, y_pred, zero_division=0),
      'Recall': recall_score(y_test_c, y_pred, zero_division=0),
      'F1': f1_score(y_test_c, y_pred, zero_division=0)
    })
class_table = pd.DataFrame(class_results).sort_values(
    ['F1', 'Accuracy'], ascending=False)
print(class_table.round(4).to_string(index=False))
best_class_name = class_table.iloc[0]['Model']
best_class_model = fitted_models[best_class_name]
y_pred_best = best_class_model.predict(X_test_c)
print('Best classifier:', best_class_name)
print('Confusion matrix:')
print(confusion_matrix(y_test_c, y_pred_best))
print(classification_report(y_test_c, y_pred_best))
```

```python
# Importing the libraries
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

#Importing the dataset and selecting features
cluster_cols = ['Hours_Studied', 'Attendance',
                'Previous_Scores', 'Tutoring_Sessions']
cluster_X = dataset[cluster_cols].copy()
cluster_X = cluster_X.fillna(cluster_X.median())
sc = StandardScaler()
cluster_scaled = sc.fit_transform(cluster_X)

#Using the Elbow Method to find the number of clusters
wcss = []
silhouettes = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++',
                    random_state=42, n_init=10)
    kmeans.fit(cluster_scaled)
    wcss.append(kmeans.inertia_)
    if 2 <= i <= 6:
        score = silhouette_score(cluster_scaled, kmeans.labels_,
                                 sample_size=1500, random_state=42)
        silhouettes.append((i, score))
plt.plot(range(1, 11), wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
print('Silhouette scores:', silhouettes)
best_k = max(silhouettes, key=lambda item: item[1])[0]
print('Selected number of clusters:', best_k)
```

```python
# d. Applying K-Means on the dataset 
kmeans = KMeans(n_clusters=best_k, init='k-means++',
                random_state=42, n_init=10)
y_kmeans = kmeans.fit_predict(cluster_scaled)
cluster_data = dataset.copy()
cluster_data['Cluster'] = y_kmeans
print(cluster_data['Cluster'].value_counts().sort_index())

# e. Visualising the clusters with PCA (as in Lab 11-12)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(cluster_scaled)
for i in range(best_k):
    plt.scatter(X_pca[y_kmeans == i, 0],
                X_pca[y_kmeans == i, 1], s=12,
                label='Cluster ' + str(i))
plt.title('Student Performance Clusters')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.show()
```
